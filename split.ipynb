```python
import numpy as np
import pandas as pd
```

```python
ori = pd.read_csv("./data_20250208.csv")
```

```python
print(ori.head(1))
```

```python
ori['date'] = pd.to_datetime(ori['index']).dt.date
ori['time'] = pd.to_datetime(ori['index']).dt.time
```

```python
# Split datatime to differenet columns, and reoder columns
ori_cols = ori.columns.tolist()
ori_cols = ['date'] + ['time'] + ori_cols[1:10]
ori = ori[ori_cols]
```

```python
OHLC = pd.DataFrame(data=ori, columns=['date', 'time', 'open', 'high', 'low', 'close', 'volume'])
DaHu = pd.DataFrame(data=ori, columns=['date', 'time', 'largeForce', 'smallForce', 'bidForce', 'askForce'])
OHLC.to_csv("./newOHLC.csv", index=False)
DaHu.to_csv("./newDaHu.csv", index=False)
```
